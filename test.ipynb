````python
class Queue:
    def __init__(self):
        self.items = []

    def isEmpty(self):
        if self.items == []:
            return True
        else:
            return False

    def enqueue(self, item):
        self.items.append(item)

    def dequeue(self):
        if len(self.items) != 0:
            return self.items.pop(0)
        else:
            print("Queue is empty")

    def display(self):
        return (self.items)

if __name__ == "__main__":
    Q = Queue()
    Q.enqueue(1)
    Q.enqueue(2)
    Q.enqueue(3)
    Q.enqueue(4)
    Q.display()
    Q.dequeue()
    Q.dequeue()
    Q.dequeue()
    Q.display()
````

````python
# 함수 생성
def queue_program():
    # 데이터를 담을 큐를 생성
    queue = list()
    while True:
        # 현재 큐에 담긴 데이터 출력
        print(queue)
        # 입력을 받는 부분
        select = input(f"1. 데이터추가\n2. 데이터 출력\nexit. 프로그램 종료\n입력: ")
        
        # 들어온 값을 공백문자를 기준으로 나눴을 때 두 개의 값으로 나뉘는지 여부
        # 문자열.split(구분자문자열)

        splited = select.split(" ")
        if len(splited) > 1:
            queue.append(splited[1]) # 데이터 추가
            continue # 다음 반복으로 이동

        elif len(splited) == 1:
            if select == "2":
                # 요소의 개수를 확인
                if len(queue) == 0:
                #   요소가 없으면 "요소가 없습니다" 출력
                    print("요소가 없습니다.")
                    continue
                else:
                    # 요소가 있으므로 0번째 요소 출력
                    print(queue.pop(0))
                    continue

            # 3. "exit"를 입력하면 프로그램 종료(break)
            # elif select == "exit":
            elif select in ["exit", "esc", "x"]: # 다중 명령문
                print("프로그램 종료")
                break # while문 탈출
            else:
                # 미리 정해진 입력이 아닌 경우
                print("알 수 없는 값입니다. 다시 입력해주세요.")

if __name__ == "__main__":
    queue_program()
````

````python
class Stack:
    def __init__(self):
        self.data = []

    # push() 구현
    def push(self, value):
        self.data.append(value) # .append를 이용하여 마지막 순서에 값 삽입
        print(f"{value} 추가됨")

    # pop() 구현
    def pop(self):
        self.popnum = self.data[-1] # 삭제된 값 출력을 위해 인덱스 마지막번째 값 저장
        del self.data[-1] # 인덱스 마지막번째 값 삭제
        print(f"{self.popnum} 삭제됨")

    # 전체 출력
    def show(self):
        print(self.data)

arr = Stack() # 객체 생성

print("==데이터 추가==")
arr.push(10)
arr.push(20)
arr.push(30)
arr.push(40)

print("==데이터 삭제==")
arr.pop()
arr.pop()
arr.pop()

print("==전체 출력==")
arr.show()
````

````python
# 링크드 리스트의 기본 단위가 되는 Node 객체를 정의합니다.
class Node:
    # 객체 생성 시 데이터 값을 매개변수로 받아 초기화합니다.
    def __init__(self, data):
        self.data = data  # Node가 저장할 실제 데이터 필드입니다.
        self.next = None  # 다음 Node의 메모리 주소를 가리키는 포인터 필드입니다. (초기값은 참조 없음)

# Node들을 연결하여 관리하는 LinkedList 클래스를 정의합니다.
class LinkedList:
    # 리스트 생성 시 시작점을 초기화합니다.
    def __init__(self):
        self.head = None  # 리스트의 첫 번째 Node를 가리키는 포인터입니다. (초기 상태는 빈 리스트)

    # [기능 1] 데이터 추가 (리스트의 맨 뒤에 삽입)
    def append(self, data):
        new_node = Node(data)  # 삽입할 데이터를 가진 새로운 Node 객체를 생성합니다.
        
        # head 포인터가 None이면 리스트가 비어있음을 의미하므로, head가 새 Node를 참조하게 합니다.
        if self.head is None:
            self.head = new_node
            return
            
        # 리스트가 비어있지 않다면, 맨 마지막 Node까지 순차 탐색을 진행해야 합니다.
        current = self.head  # 탐색의 시작점을 head로 설정합니다.
        
        # 현재 Node의 next 포인터가 None이 아닐 때까지(다음 Node가 존재할 때까지) 반복합니다.
        while current.next is not None:
            current = current.next  # 참조를 이동하여 다음 Node로 접근합니다.
            
        # loop를 빠져나왔을 때 current는 마지막 Node입니다. 
        # 마지막 Node의 next 포인터가 새로 생성한 Node를 가리키도록 주소를 할당합니다.
        current.next = new_node

    # [기능 2] 데이터 탐색 (순차 검색)
    def search(self, target):
        current = self.head  # 탐색의 시작점을 첫 번째 Node(head)로 설정합니다.
        
        # current가 None이 될 때까지(리스트의 끝에 도달할 때까지) Node를 하나씩 방문합니다.
        while current is not None:
            # 현재 Node의 데이터 필드 값이 찾고자 하는 대상(target)과 일치하는지 비교합니다.
            if current.data == target:
                return True  # 일치하는 데이터를 찾은 경우 탐색을 종료하고 True를 반환합니다.
            
            current = current.next  # 일치하지 않으면 next 포인터를 따라 다음 Node로 주소를 이동합니다.
            
        return False  # 리스트의 마지막까지 탐색했으나 일치하는 데이터가 없는 경우 False를 반환합니다.

    # [기능 3] 데이터 삭제 (특정 값을 가진 Node 참조 제거)
    def delete(self, target):
        # 리스트가 비어있는 상태(head가 None)라면 삭제 연산을 수행하지 않고 종료합니다.
        if self.head is None:
            return

        # 삭제하고자 하는 데이터가 head Node(첫 번째 Node)에 위치한 경우입니다.
        if self.head.data == target:
            # head 포인터가 다음 Node(head.next)를 가리키도록 변경하여 첫 번째 Node의 참조를 끊습니다.
            self.head = self.head.next  
            return

        # 삭제할 데이터가 중간이나 끝에 위치한 경우를 위한 탐색입니다.
        current = self.head  # head부터 탐색을 시작합니다.
        
        # 현재 Node의 다음 Node(current.next)가 존재하는 동안 반복합니다.
        while current.next is not None:
            # '다음 Node'의 데이터가 삭제 대상(target)인지 확인합니다.
            if current.next.data == target:
                # 현재 Node의 next 포인터가 삭제 대상 Node를 건너뛰고,
                # 삭제 대상 Node의 '다음 Node(current.next.next)'를 직접 가리키도록 주소를 바꿉니다.
                current.next = current.next.next
                return  # 참조 변경이 완료되면 연산을 종료합니다.
                
            current = current.next  # 대상을 찾지 못했다면 다음 Node로 이동하여 검사를 계속합니다.

    # [기능 4] 전체 리스트 출력 (시각화용)
    def display(self):
        current = self.head
        nodes = []
        # head부터 순차적으로 접근하여 각 Node의 data 값을 문자열로 변환 후 배열에 저장합니다.
        while current is not None:
            nodes.append(str(current.data))
            current = current.next
        # 저장된 데이터들을 화살표 기호로 연결하여 출력 포맷을 구성합니다.
        print(" -> ".join(nodes) + " -> None")

# 1. 링크드 리스트 생성
my_list = LinkedList()

# 2. 데이터 추가 (Append)
my_list.append(10)
my_list.append(20)
my_list.append(30)
print("데이터 추가 후:")
my_list.display()  # 출력: 10 -> 20 -> 30 -> None

# 3. 데이터 탐색 (Search)
print("\n데이터 탐색:")
print("20이 리스트에 있나요?", my_list.search(20))  # 출력: True
print("40이 리스트에 있나요?", my_list.search(40))  # 출력: False

# 4. 데이터 삭제 (Delete)
my_list.delete(20)
print("\n20 삭제 후:")
my_list.display()  # 출력: 10 -> 30 -> None
````

````python
def fractional_knapsack(items, capacity):
    # 각 보석에 1kg당 가치 정보를 추가한다.
    for item in items:
        name, weight, value = item
        value_per_weight = value / weight
        item.append(value_per_weight)

    # 1kg당 가치가 높은 순서대로 정렬한다.
    items.sort(key=lambda item: item[3], reverse=True)

    total_value = 0
    remaining_capacity = capacity

    for name, weight, value, value_per_weight in items:
        # 배낭이 가득 찼다면 반복을 종료한다.
        if remaining_capacity == 0:
            break

        # 보석을 전부 담을 수 있는 경우
        if weight <= remaining_capacity:
            remaining_capacity -= weight
            total_value += value

            print(f"보석 {name}을(를) {weight}kg 담았습니다.")

        # 보석의 일부만 담을 수 있는 경우
        else:
            selected_weight = remaining_capacity
            selected_value = selected_weight * value_per_weight

            total_value += selected_value
            remaining_capacity = 0

            print(f"보석 {name}을(를) {selected_weight}kg 담았습니다.")

    return total_value


items = [
    ["A", 5, 50],
    ["B", 10, 60],
    ["C", 8, 80],
    ["D", 4, 40]
]

capacity = 15

result = fractional_knapsack(items, capacity)

print(f"최대 가치는 {int(result)}만원입니다.")
````

````python
import heapq
import time


# 무방향 가중치 그래프
graph = {
    "A": [("B", 2), ("C", 5)],
    "B": [("A", 2), ("C", 1), ("D", 3), ("E", 4)],
    "C": [("A", 5), ("B", 1), ("D", 2), ("F", 6)],
    "D": [("B", 3), ("C", 2), ("E", 1), ("F", 2)],
    "E": [("B", 4), ("D", 1), ("F", 3)],
    "F": [("C", 6), ("D", 2), ("E", 3)]
}

# heapq를 사용한 다익스트라 알고리즘
def dijkstra_heap(graph, start):
    # 시작 정점에서 각 정점까지의 최단 거리
    distances = {
        vertex: float("inf")
        for vertex in graph
    }

    # 각 정점에 도착하기 직전의 정점
    previous = {
        vertex: None
        for vertex in graph
    }

    # 시작 정점의 거리는 0
    distances[start] = 0

    # (거리, 정점) 형태로 우선순위 큐에 저장
    queue = [(0, start)]

    while queue:
        # 현재 가장 거리가 짧은 정점을 꺼낸다.
        current_distance, current_vertex = heapq.heappop(queue)

        # 이미 더 짧은 경로를 찾은 정점이면 건너뛴다.
        if current_distance > distances[current_vertex]:
            continue

        # 현재 정점과 연결된 정점들을 확인한다.
        for neighbor, weight in graph[current_vertex]:
            new_distance = current_distance + weight

            # 새로운 경로가 기존 경로보다 짧다면 갱신한다.
            if new_distance < distances[neighbor]:
                distances[neighbor] = new_distance
                previous[neighbor] = current_vertex

                heapq.heappush(
                    queue,
                    (new_distance, neighbor)
                )

    return distances, previous


# --------------------------------------------------
# 실제 최단 경로 구하기
# --------------------------------------------------
def get_path(previous, start, end):
    path = []
    current = end

    # 도착 정점에서 이전 정점을 따라 거꾸로 이동한다.
    while current is not None:
        path.append(current)
        current = previous[current]

    # 현재 경로는 도착 → 시작 순서이므로 뒤집는다.
    path.reverse()

    # 시작 정점까지 연결되지 않은 경우
    if not path or path[0] != start:
        return []

    return path


# --------------------------------------------------
# heapq를 사용하지 않은 다익스트라 알고리즘
# --------------------------------------------------
def dijkstra_without_heap(graph, start):
    distances = {
        vertex: float("inf")
        for vertex in graph
    }

    previous = {
        vertex: None
        for vertex in graph
    }

    visited = set()
    distances[start] = 0

    while len(visited) < len(graph):
        current_vertex = None
        current_distance = float("inf")

        # 방문하지 않은 정점 중 거리가 가장 짧은 정점을 찾는다.
        for vertex in graph:
            if (
                vertex not in visited
                and distances[vertex] < current_distance
            ):
                current_vertex = vertex
                current_distance = distances[vertex]

        # 더 이상 방문할 수 있는 정점이 없는 경우
        if current_vertex is None:
            break

        visited.add(current_vertex)

        # 현재 정점과 연결된 정점의 거리를 확인한다.
        for neighbor, weight in graph[current_vertex]:
            new_distance = (
                distances[current_vertex] + weight
            )

            if new_distance < distances[neighbor]:
                distances[neighbor] = new_distance
                previous[neighbor] = current_vertex

    return distances, previous


# --------------------------------------------------
# 결과 출력 함수
# --------------------------------------------------
def print_result(distances, previous, start):
    print(f"\n시작 정점: {start}")
    print("-" * 45)

    for vertex in graph:
        path = get_path(previous, start, vertex)

        if distances[vertex] == float("inf"):
            print(f"{vertex}: 이동할 수 없음")

        else:
            path_text = " → ".join(path)

            print(
                f"{vertex}: "
                f"최단 거리 = {distances[vertex]}, "
                f"경로 = {path_text}"
            )


# --------------------------------------------------
# 프로그램 실행
# --------------------------------------------------
start = "A"

heap_distances, heap_previous = dijkstra_heap(
    graph,
    start
)

print("heapq를 사용한 다익스트라 결과")
print_result(
    heap_distances,
    heap_previous,
    start
)


normal_distances, normal_previous = (
    dijkstra_without_heap(graph, start)
)

print("\nheapq를 사용하지 않은 다익스트라 결과")
print_result(
    normal_distances,
    normal_previous,
    start
)


# --------------------------------------------------
# 실행 시간 비교
# --------------------------------------------------
repeat = 10000

heap_start = time.perf_counter()

for _ in range(repeat):
    dijkstra_heap(graph, start)

heap_end = time.perf_counter()


normal_start = time.perf_counter()

for _ in range(repeat):
    dijkstra_without_heap(graph, start)

normal_end = time.perf_counter()


heap_time = heap_end - heap_start
normal_time = normal_end - normal_start


print("\n성능 비교")
print("-" * 45)
print(f"반복 횟수: {repeat:,}회")
print(f"heapq 사용 시간     : {heap_time:.6f}초")
print(f"heapq 미사용 시간   : {normal_time:.6f}초")

if heap_time < normal_time:
    print(
        f"heapq 방식이 약 "
        f"{normal_time / heap_time:.2f}배 빠릅니다."
    )
else:
    print(
        "그래프가 작아서 heapq 방식의 차이가 "
        "크게 나타나지 않았습니다."
    )
````
